```python
import pandas as pd
df=pd.read_csv('netflix_titles.csv.zip')
df.head()
df.columns
df.info()
df.describe()
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(14,6))
sns.countplot(x='release_year',hue='type',data=df[df['release_year']>=2000],palette='dark')
plt.xticks(rotation=45)
plt.xlabel('Release Year')
plt.ylabel('Number of Shows')
plt.title('Number of Shows released in a year')
plt.grid(axis='y',linestyle='--',alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
valid_ratings=['TV-MA','TV-14','TV-PG','R','PG-13','TV-Y7','TV-Y','PG','TV-G','NR','G','UR','NC-17']
df_cleaned=df[df['rating'].isin(valid_ratings)]
plt.figure(figsize=(14,7))
sns.countplot(x='rating',data=df_cleaned,color='maroon',order=df_cleaned['rating'].value_counts().index)
plt.title('Number of shows by Rating')
plt.xlabel('Rating')
plt.ylabel('Number of shows')
plt.xticks(rotation=45)
plt.grid(axis='y',linestyle='--',alpha=0.3)
plt.text(x=10,y=plt.ylim()[1]*0.85,s='TV-MA:Mature 17+\nTV-14:Age 14+\nTV-PG:Parental Guidance\nTV-Y:All Children\nG:General Audience',fontsize=9,bbox=dict(facecolor='none',edgecolor='black',boxstyle='round,pad=0.5'))
plt.tight_layout()
plt.show()
```
